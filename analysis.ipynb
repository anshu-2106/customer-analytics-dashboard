```python
import pandas as pd

customers = pd.read_csv("Data/retail_analytics/olist_customers_dataset.csv")
orders = pd.read_csv("Data/retail_analytics/olist_orders_dataset.csv")
payments = pd.read_csv("Data/retail_analytics/olist_order_payments_dataset.csv")
items = pd.read_csv("Data/retail_analytics/olist_order_items_dataset.csv")

print("Customers:", customers.shape)
print("Orders:", orders.shape)
print("Payments:", payments.shape)
print("Items:", items.shape)
```

```python
import matplotlib.pyplot as plt

payment_revenue = payments.groupby("payment_type")["payment_value"].sum()

payment_revenue.plot(kind="bar")

plt.title("Revenue by Payment Type")
plt.xlabel("Payment Type")
plt.ylabel("Revenue")
plt.show()
```

```python
customer_orders = customers.merge(
    orders,
    on="customer_id",
    how="inner"
)

customer_orders.head()
```

```python
customer_orders.shape
```

```python
customer_orders_payments = customer_orders.merge(
    payments,
    on="order_id",
    how="inner"
)

customer_orders_payments.shape
```

```python
state_revenue = customer_orders_payments.groupby(
    "customer_state"
)["payment_value"].sum().sort_values(ascending=False)

state_revenue.head(10)
```

```python
import matplotlib.pyplot as plt

state_revenue.head(10).plot(kind="bar")

plt.title("Top 10 States by Revenue")
plt.xlabel("State")
plt.ylabel("Revenue")
plt.show()
```

```python
customer_orders_payments["order_purchase_timestamp"] = pd.to_datetime(
    customer_orders_payments["order_purchase_timestamp"]
)

customer_orders_payments["month"] = (
    customer_orders_payments["order_purchase_timestamp"]
    .dt.to_period("M")
)

monthly_revenue = customer_orders_payments.groupby(
    "month"
)["payment_value"].sum()

monthly_revenue.head()
```

```python
monthly_revenue.plot(figsize=(10,5))

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```
